# 📊 GitHub Profile Stats & Visualization Generator

This interactive notebook allows you to inspect, customize, and preview all SVG UI cards generated for your GitHub Profile README.

### 🛠 Modules & Tools
- **Environment**: 
- **Package**:  (modular rendering & GraphQL API fetching)
- **Output Path**: 

```python
import os
import sys
try:
    from IPython.display import SVG, display, HTML
except ImportError:
    SVG, display, HTML = None, print, print

# Ensure local scripts directory is in sys.path
sys.path.insert(0, os.path.abspath('.'))

from stats_generator import (
    USERNAME,
    fetch_contributions_and_user,
    fetch_all_repo_language_sizes,
    compute_streaks,
    render_header_banner,
    render_profile_overview,
    render_top_langs,
    render_streaks,
    render_activity_radar,
    render_productive_time,
    display_svg
)

print(f"Loaded stats_generator for user: {USERNAME}")
```

## 📡 1. Fetch User Data & Contribution Metrics

```python
# Fetch data via GitHub GraphQL API
user = fetch_contributions_and_user(USERNAME)
total_contribs = user['contributionsCollection']['contributionCalendar']['totalContributions']
contrib_calendar = user['contributionsCollection']['contributionCalendar']
streak_info = compute_streaks(contrib_calendar)

total_repos, languages, stars, followers_count = fetch_all_repo_language_sizes(USERNAME)
followers = user.get('followers', {}).get('totalCount', followers_count)
days = [d for w in contrib_calendar['weeks'] for d in w['contributionDays']]

print(f"User: {user.get('login')}")
print(f"Contributions: {total_contribs} | Public Repos: {total_repos} | Stars: {stars} | Followers: {followers}")
print("Top Languages:", list(sorted(languages.items(), key=lambda x: x[1], reverse=True))[:5])
```

## 🎨 2. Inspect & Preview Header Banner Card

```python
header_svg = render_header_banner(user, total_contribs, total_repos)
display_svg(header_svg)
```

## 📊 3. Inspect Profile Overview & Top Languages Cards

```python
overview_svg = render_profile_overview(user, total_contribs, total_repos, stars, followers)
langs_svg = render_top_langs(languages)

if HTML:
    display(HTML("<h3>Profile Metrics Overview</h3>"))
display_svg(overview_svg)
if HTML:
    display(HTML("<h3>Top Programming Languages</h3>"))
display_svg(langs_svg)
```

## 🔥 4. Inspect Contribution Streak Banner

```python
streak_svg = render_streaks(streak_info)
display_svg(streak_svg)
```

## 🎯 5. Inspect Skill Radar Matrix & Activity Breakdown Cards

```python
radar_svg = render_activity_radar(languages)
time_svg = render_productive_time(days)

if HTML:
    display(HTML("<h3>Skill Radar Matrix</h3>"))
display_svg(radar_svg)
if HTML:
    display(HTML("<h3>Activity Breakdown (Weekday Histogram)</h3>"))
display_svg(time_svg)
```

## 💾 6. Export & Batch Save All SVG Assets

```python
os.makedirs('assets', exist_ok=True)

render_header_banner(user, total_contribs, total_repos, out='assets/header-banner.svg')
render_profile_overview(user, total_contribs, total_repos, stars, followers, out='assets/profile-overview.svg')
render_top_langs(languages, out='assets/top-langs.svg')
render_streaks(streak_info, out='assets/streak.svg')
render_activity_radar(languages, out='assets/activity-radar.svg')
render_productive_time(days, out='assets/productive-time.svg')

print("Successfully rendered and updated all SVG cards in assets/!")
```
